```python
import matplotlib.pyplot as plt
import numpy as np
from myfuncs import plot as yplt
%matplotlib ipympl
```

```python
parent_dir = './'
table = np.load(f'{parent_dir}cib_lookup_table.npz')
```

```python
A_real = table['A'].flatten()
nu_o_real = table['nu_o'].flatten()

z_real = table['z']
```

```python
z_interp = np.linspace(0,10, int(1e6))

A_interp = np.interp(z_interp, z_real, A_real)
nu_o_interp = np.interp(z_interp, z_real, nu_o_real)
```

```python
ax = plt.figure().add_subplot(projection= '3d')

interval = int(len(z_real) / 10)

ax.plot(nu_o_real[::interval], A_real[::interval], z_real[::interval], '.', label='exact data')
ax.plot(nu_o_interp, A_interp, z_interp, label='interpolated data')

ax.set_xlabel(r'$\nu_o$')
ax.set_ylabel('A')
ax.set_zlabel('z')

plt.legend();

plt.show()
```
